```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/onlinefoods.csv')

print("Dataset berhasil dibaca!")
```

```python
df.head()
```

```python
df.info()
```

```python
# Mengecek data kosong
print("Data kosong:")
print(df.isnull().sum())

# Mengecek data duplikat
print("\nData duplikat:")
print(df.duplicated().sum())

# Menghapus kolom duplikat
df = df.drop(columns=['Unnamed: 12'])

# Membersihkan spasi Feedback
df['Feedback'] = df['Feedback'].str.strip()

# Menghapus baris duplikat
df = df.drop_duplicates()

# Menampilkan hasil
print("\nData setelah dibersihkan:")
print(df.shape)

print("\nCek duplikat:")
print(df.duplicated().sum())
```

```python
# ANALISIS STATISTIK DESKRIPTIF

print("===== STATISTIK USIA =====")

print("Mean:", df['Age'].mean())
print("Median:", df['Age'].median())
print("Modus:", df['Age'].mode()[0])
print("Standar Deviasi:", df['Age'].std())
print("Variansi:", df['Age'].var())
print("Nilai Minimum:", df['Age'].min())
print("Nilai Maksimum:", df['Age'].max())
print("Range:", df['Age'].max() - df['Age'].min())

print("\n===== STATISTIK UKURAN KELUARGA =====")

print("Mean:", df['Family size'].mean())
print("Median:", df['Family size'].median())
print("Modus:", df['Family size'].mode()[0])
print("Standar Deviasi:", df['Family size'].std())

print("\n===== KUARTIL USIA =====")

print("Q1:", df['Age'].quantile(0.25))
print("Q2:", df['Age'].quantile(0.50))
print("Q3:", df['Age'].quantile(0.75))

print("\n===== KATEGORI RESPONDEN =====")

print(df['Occupation'].value_counts())

print("\n===== HASIL PEMESANAN ONLINE =====")

print(df['Output'].value_counts())

print("\n===== FEEDBACK PELANGGAN =====")

print(df['Feedback'].value_counts())
```
